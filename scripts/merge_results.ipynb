```python
import json

with open('../data/eval_results.json') as f:
    source = {s['name']: s for s in json.load(f)}

with open('../data/eval_results_results.json') as f:
    labeled = json.load(f)

labeled_lookup = {s['name']: s for s in labeled}

updated = 0
for session in labeled:
    name = session['name']
    if name in source and not session.get('result'):
        session['result'] = source[name].get('result', '')
        session['status'] = source[name].get('status', 'error')
        updated += 1

with open('../data/eval_results_results.json', 'w') as f:
    json.dump(labeled, f, indent=2)

print(f"Updated {updated} sessions")
```

```python
import json

with open('../data/eval_results.json') as f:
    source = {s['name']: s for s in json.load(f)}

with open('../data/eval_results_results.json') as f:
    labeled = {s['name']: s for s in json.load(f)}

print(f"eval_results.json: {len(source)} sessions")
print(f"eval_results_results.json: {len(labeled)} sessions")

print("\nMismatches in result field:")
mismatches = 0
for name in source:
    if name in labeled:
        src_result = source[name].get('result', '')
        lab_result = labeled[name].get('result', '')
        if src_result != lab_result and source[name].get('status') == 'success':
            print(f"  {name}: results differ")
            mismatches += 1

print(f"\nTotal mismatches: {mismatches}")
print(f"\nMissing from labeled file:")
for name in source:
    if name not in labeled:
        print(f"  {name}")
```

```python
import json
with open('../data/eval_results_results.json') as f:
    data = json.load(f)
labeled = [s for s in data if s.get('label') is not None]
unlabeled_with_results = [s for s in data if s.get('label') is None and s.get('result') and 'ERROR' not in str(s.get('result',''))]
errors = [s for s in data if s.get('label') is None and 'ERROR' in str(s.get('result',''))]
no_result = [s for s in data if not s.get('result')]
print(f'Total: {len(data)}')
print(f'Labeled: {len(labeled)}')
print(f'Unlabeled with results (need labeling): {len(unlabeled_with_results)}')
print(f'Unlabeled errors: {len(errors)}')
print(f'No result yet: {len(no_result)}')
```
